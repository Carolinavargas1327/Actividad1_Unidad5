```python
import transformers
import torch
import streamlit
import pandas
import plotly

print("Transformers:", transformers.__version__)
print("PyTorch:", torch.__version__)
print("Streamlit:", streamlit.__version__)
print("Pandas:", pandas.__version__)
print("Plotly:", plotly.__version__)
```

```python
from transformers import pipeline
import streamlit as st
import pandas as pd
import plotly.express as px
```

```python
analizador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis"
)
```

```python
texto = st.text_area(
    "Escribe un comentario:",
    placeholder="Ejemplo: Me encanta esta aplicación."
)
```

```python
if st.button("Analizar sentimiento"):
    if texto.strip():
        resultado = analizador(texto)[0]
        etiqueta = resultado["label"]
        confianza = resultado["score"]
```

```python
texto = "Me encanta esta aplicación"

resultado = analizador(texto)[0]

etiqueta = resultado["label"]
confianza = resultado["score"]

print("Sentimiento:", etiqueta)
print("Confianza:", f"{confianza:.2%}")
```

```python
st.metric("Sentimiento identificado", etiqueta)
st.metric("Nivel de confianza", f"{confianza:.2%}")
```
