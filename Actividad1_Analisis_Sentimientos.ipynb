```python
from transformers import pipeline
import streamlit as st
import pandas as pd
import plotly.express as px

print("Bibliotecas importadas correctamente")
```

```python
analizador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis"
)

print("Modelo cargado correctamente")
```

```python
textos = [
    "Me encanta esta aplicación, es excelente.",
    "El servicio es terrible, no me gustó.",
    "Hoy fui a la escuela y tuve clases."
]

for texto in textos:
    resultado = analizador(texto)

    print("Texto:", texto)
    print("Sentimiento:", resultado[0]["label"])
    print("Confianza:", resultado[0]["score"])
    print("-------------------------")
```

```python
textos_especiales = [
    "¡Qué maravilloso, otra vez se fue el internet!",
    "No estuvo tan mal la película.",
    "Me encanta esperar tres horas para recibir atención."
]

for texto in textos_especiales:
    resultado = analizador(texto)

    print("Texto:", texto)
    print("Sentimiento:", resultado[0]["label"])
    print("Confianza:", resultado[0]["score"])
    print("-------------------------")
```
